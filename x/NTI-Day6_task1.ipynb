```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("manarm7md/maqa-dataset")

print("Path to dataset files:", path)
```

```python
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score
)

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Embedding,
    SimpleRNN,
    Dense,
    Dropout
)

from tensorflow.keras.callbacks import EarlyStopping
```

```python
train = pd.read_excel(path + "/Balanced/Balanced/Train.xlsx")
test= pd.read_excel(path + "/Balanced/Balanced/Test.xlsx")
```

```python
classes = [
    "الطب العام",
    "امراض الاطفال",
    "امراض الجهاز التنفسي",
    "امراض الجهاز الهضمي",
    "امراض القلب و الشرايين"
]
```

```python
df_train = train[train['category'].isin(classes)].copy()
df_test = test[test['category'].isin(classes)].copy()

print("Train:", df_train.shape)
print("Test :", df_test.shape)

print("\nTrain distribution:")
print(df_train['category'].value_counts())

print("\nTest distribution:")
print(df_test['category'].value_counts())
```

```python
import re
import pandas as pd


def clean_arabic_text(text):
    text = str(text)

    # =========================
    # 1. Remove URLs
    # =========================
    text = re.sub(r'https?://\S+|www\.\S+', ' ', text)

    # =========================
    # 2. Remove HTML
    # =========================
    text = re.sub(r'<[^>]+>', ' ', text)

    # =========================
    # 3. Remove Tashkeel
    # =========================
    text = re.sub(
        r'[\u0617-\u061A\u064B-\u065F\u0670]',
        '',
        text
    )

    # =========================
    # 4. Normalize Arabic letters
    # =========================

    # Alef variations
    text = re.sub(r'[إأآٱ]', 'ا', text)

    # Alef Maqsura
    text = text.replace('ى', 'ي')

    # Remove Tatweel
    text = text.replace('ـ', '')

    # =========================
    # 5. Remove punctuation
    # Keep Arabic + English + numbers
    # =========================
    text = re.sub(r'[^\w\s]', ' ', text)

    # =========================
    # 6. Remove English letters
    # =========================
    text = re.sub(r'[A-Za-z]', ' ', text)

    # =========================
    # 7. Normalize numbers
    # =========================
    text = re.sub(r'\d+', ' ', text)

    # =========================
    # 8. Remove extra spaces
    # =========================
    text = re.sub(r'\s+', ' ', text).strip()

    return text


# ==========================================
# Apply preprocessing
# ==========================================

df_train['q_clean'] = df_train['q_body'].apply(clean_arabic_text)
df_train['a_clean'] = df_train['a_body'].apply(clean_arabic_text)

df_test['q_clean'] = df_test['q_body'].apply(clean_arabic_text)
df_test['a_clean'] = df_test['a_body'].apply(clean_arabic_text)


# ==========================================
# Combine Question + Answer
# ==========================================

df_train['text'] = (
    'سؤال ' +
    df_train['q_clean'] +
    ' جواب ' +
    df_train['a_clean']
)

df_test['text'] = (
    'سؤال ' +
    df_test['q_clean'] +
    ' جواب ' +
    df_test['a_clean']
)


# ==========================================
# Check results
# ==========================================

print(df_train[['q_body', 'q_clean', 'a_body', 'a_clean', 'category']]
      .head(5)
      .to_string(index=False))
```

```python
label_encoder = LabelEncoder()

df_train['label'] = label_encoder.fit_transform(
    df_train['category']
)

df_test['label'] = label_encoder.transform(
    df_test['category']
)

print("Classes:")
for i, cls in enumerate(label_encoder.classes_):
    print(i, "->", cls)
```

```python
MAX_WORDS = 30000

tokenizer = Tokenizer(
    num_words=MAX_WORDS,
    oov_token='<OOV>'
)

tokenizer.fit_on_texts(df_train['text'])
```

```python
X_train = tokenizer.texts_to_sequences(
    df_train['text']
)

X_test = tokenizer.texts_to_sequences(
    df_test['text']
)
```

```python

lengths = [len(x) for x in X_train]

print("Max length:", max(lengths))
print("Mean length:", np.mean(lengths))
print("95th percentile:", np.percentile(lengths, 95))

MAX_LEN = int(np.percentile(lengths, 95))

print("MAX_LEN:", MAX_LEN)
```

```python
MAX_LEN = int(np.percentile(lengths, 95))

print("MAX_LEN:", MAX_LEN)
```

```python
MAX_LEN = 56
```

```python
X_train = pad_sequences(
    X_train,
    maxlen=MAX_LEN,
    padding='post',
    truncating='post'
)

X_test = pad_sequences(
    X_test,
    maxlen=MAX_LEN,
    padding='post',
    truncating='post'
)

print(X_train.shape)
print(X_test.shape)
```

```python
y_train = df_train['label'].values
y_test = df_test['label'].values

print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
X_train, X_val, y_train, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.15,
    random_state=42,
    stratify=y_train
)

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_val:", X_val.shape)
print("y_val:", y_val.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
vocab_size = min(
    MAX_WORDS,
    len(tokenizer.word_index) + 1
)

print("Vocabulary size:", vocab_size)
```

#RNN

```python
num_classes = len(label_encoder.classes_)

model = Sequential([

    Embedding(
        input_dim=vocab_size,
        output_dim=128
    ),

    SimpleRNN(
        128,
        return_sequences=False
    ),


    Dense(
        64,
        activation='relu'
    ),

    Dense(
        num_classes,
        activation='softmax'
    )
])
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True
)
```

```python
import time

# =========================
# RNN Runtime
# =========================

rnn_start = time.time()

history = model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=15,
    batch_size=64,
    callbacks=[early_stop],
    verbose=1
)

rnn_runtime = time.time() - rnn_start

print(f"RNN Runtime: {rnn_runtime:.2f} seconds")
```

```python
test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=1
)

print(f"Test Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_accuracy:.4f}")
```

```python
y_pred_prob = model.predict(X_test)

y_pred = np.argmax(
    y_pred_prob,
    axis=1
)
```

```python
print(
    classification_report(
        y_test,
        y_pred,
        target_names=label_encoder.classes_
    )
)
```

#CNN

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Embedding,
    Conv1D,
    GlobalMaxPooling1D,
    Dense,
    Dropout
)

cnn_num_classes = len(label_encoder.classes_)

cnn_model = Sequential([

    Embedding(
        input_dim=vocab_size,
        output_dim=128
    ),

    Conv1D(
        filters=128,
        kernel_size=3,
        activation='relu'
    ),

    GlobalMaxPooling1D(),

    Dense(
        128,
        activation='relu'
    ),

    Dropout(0.4),

    Dense(
        cnn_num_classes,
        activation='softmax'
    )
])

cnn_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

cnn_model.summary()
```

```python
cnn_early_stop = EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True
)
# =========================
# CNN Runtime
# =========================

cnn_start = time.time()

cnn_history = cnn_model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=15,
    batch_size=64,
    callbacks=[cnn_early_stop],
    verbose=1
)

cnn_runtime = time.time() - cnn_start

```

```python
cnn_test_loss, cnn_test_accuracy = cnn_model.evaluate(
    X_test,
    y_test,
    verbose=1
)

print(f"CNN Test Loss: {cnn_test_loss:.4f}")
print(f"CNN Test Accuracy: {cnn_test_accuracy:.4f}")
```

```python
cnn_y_pred_prob = cnn_model.predict(X_test)

cnn_y_pred = np.argmax(
    cnn_y_pred_prob,
    axis=1
)
```

#BIRNN

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Embedding,
    SimpleRNN,
    Bidirectional,
    Dense,
    Dropout
)

birnn_model = Sequential([

    Embedding(
        input_dim=vocab_size,
        output_dim=128
    ),

    Bidirectional(
        SimpleRNN(
            128,
            return_sequences=False
        )
    ),

    Dropout(0.3),

    Dense(
        64,
        activation='relu'
    ),

    Dropout(0.3),

    Dense(
        len(label_encoder.classes_),
        activation='softmax'
    )
])

birnn_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

birnn_model.summary()
```

```python
birnn_early_stop = EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True
)
```

```python
import time

birnn_start = time.time()

birnn_history = birnn_model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=15,
    batch_size=64,
    callbacks=[birnn_early_stop],
    verbose=1
)

birnn_runtime = time.time() - birnn_start

print(f"BiRNN Runtime: {birnn_runtime:.2f} seconds")
print(f"BiRNN Runtime: {birnn_runtime/60:.2f} minutes")
```

```python
birnn_test_loss, birnn_test_accuracy = birnn_model.evaluate(
    X_test,
    y_test,
    verbose=1
)

print(f"BiRNN Test Loss: {birnn_test_loss:.4f}")
print(f"BiRNN Test Accuracy: {birnn_test_accuracy:.4f}")
```

```python
birnn_y_pred_prob = birnn_model.predict(X_test)

birnn_y_pred = np.argmax(
    birnn_y_pred_prob,
    axis=1
)
```
